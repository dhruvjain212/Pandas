```python
#Reading data from a csv file
obesity_data = pd.read_csv('./obesity_pandas_practice.csv')
obesity_data.head() #to display the first 5 rows of the dataframe
obesity_data.tail() #to display the last 5 rows of the dataframe, write obesity_data.tail(10) to display the last 10 rows of the dataframe
obesity_data.info() #to display the summary of the dataframe
```

```python
#To generate random data from sample
obesity_data.sample()
obesity_data.sample(5) #to display 5 random rows of the dataframe, write obesity_data.sample(10) to display 10 random rows of the dataframe

#to fix the state of data to generate same random data every time, we can use the random_state parameter
obesity_data.sample(5, random_state = 1) #to display 5 random rows of the dataframe, write obesity_data.sample(10, random_state = 1) to display 10 random rows of the dataframe
```

```python
#converting columns to a list
columns_list = obesity_data.columns.tolist() #to convert the columns of the dataframe to a list
print(columns_list) #to display the list of columns of the dataframe
print(list(obesity_data.index)) #to display the list of indices of the dataframe
```
