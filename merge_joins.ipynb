### Merge, joins and concatenate

```python
import pandas as pd
```

```python
ob_data = pd.read_csv('./obesity_pandas_practice.csv')
```

```python
df1 = pd.DataFrame({
    'id' : [1,2,3,4],
    'name': ['Dhruv', 'Amit', 'Educosys', 'Youtube']
})

df2 = pd.DataFrame({
    'id': [3,4,5,6],
    'course': ['GenAI', 'CPP', 'HLD', 'LLD']
})
```

```python
inner_result = pd.merge(df1, df2, how = 'inner', on = 'id') # inner join is like intersection
inner_result
```

```python
outer_result = pd.merge(df1, df2, how = 'outer', on = 'id') # outer join is like union
outer_result
```

```python
# when the name of column is different in both the dataframes
df3 = pd.DataFrame({
    'id1': [1,2,3,4],
    'name': ['Dhruv', 'Amit', 'Educosys', 'Youtube']
})

df4 = pd.DataFrame({
    'id2': [3,4,5,6],
    'course': ['GenAI', 'CPP', 'HLD', 'LLD']
})

inner_result = pd.merge(df3, df4, how = 'inner', left_on = 'id1', right_on = 'id2')
inner_result
```
