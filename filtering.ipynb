### Data Filtering Methods

```python
import pandas as pd
```

```python
ob_data = pd.read_csv('./obesity_pandas_practice.csv')
```

```python
ob_data.loc[ob_data['Weight'] < 50]
```

```python
ob_data.loc[(ob_data['Weight'] < 50) & (ob_data['Obesity'] == 'Normal_Weight')]
```

```python
ob_data.loc[ob_data['Obesity'].str.contains('Normal')]
```

```python
ob_data.loc[ob_data['Obesity'].str.startswith('Nor')]
```
