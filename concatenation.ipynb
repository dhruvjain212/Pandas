### Concatenation of two dataframes


```python
import pandas as pd
```

```python
# when the name of column is different in both the dataframes
df1 = pd.DataFrame({
    'id1': [1,2,3,4],
    'name': ['Dhruv', 'Amit', 'Educosys', 'Youtube']
})

df2 = pd.DataFrame({
    'id2': [3,4,5,6],
    'course': ['GenAI', 'CPP', 'HLD', 'LLD']
})

inner_result = pd.concat([df1, df2], axis = 0)
inner_result
```

```python
inner_result = pd.concat([df1, df2], axis = 0, ignore_index = True) #ignore_index is used to keep the id sequentially incremented
inner_result
```

```python
result = pd.concat([df1, df2], axis = 0, keys = ['df1', 'df2'])
result
```

```python
result2 = pd.concat([df1, df2], axis = 1)
result2
```
