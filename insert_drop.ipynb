### Insert and Drop Methods
    

```python
import pandas as pd
```

```python
ob_data = pd.read_csv('./obesity_pandas_practice.csv')
```

```python
ob_data.insert(2, 'BMI', 20)
```

```python
ob_data.head()
```

```python
ob_data.drop(columns = 'BMI') # doesnot change the dataframe
ob_data.head()

#to change the dataframe
ob_data = ob_data.drop(columns = 'BMI')
ob_data.head()

#OR
ob_data.drop(columns = 'BMI', inplace = True)
```
