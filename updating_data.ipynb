### Updating data

```python
import pandas as pd
```

```python
ob_data = pd.read_csv('./obesity_pandas_practice.csv')
```

```python
ob_data.loc[0, 'Age'] = 22
ob_data.head()
```

```python
ob_data.loc[:, 'SMOKE'] = 'yes'
ob_data.head()
```

```python
ob_data.iloc[:, 9] = 'doesnt matter'
ob_data.head()
```

```python
ob_data.at[2, 'Obesity'] = 'Not Obese'
ob_data.head()
```

```python
ob_data.iat[2,0] = 'Secret'
ob_data.head()
```

```python
ob_data['SMOKE'] = 'Yes'
ob_data.head()
```

```python
#Updating the column of each row
def increment_by_5(x):
    return x+5;

ob_data['Age'] = ob_data['Age'].apply(increment_by_5)
ob_data.head()
```

```python
#using lambda function

ob_data['Age'] = ob_data['Age'].apply(lambda x : x-5);
ob_data.head()
```

```python
ob_data['Age Category'] = ob_data['Age'].apply(lambda x: 'Very young' if x <25 else 'Mature')
ob_data.head()
```

```python
import numpy as np
```

```python
ob_data['New Age Category'] = np.where(ob_data['Age'] < 25, 'Very Young', 'Mature')
ob_data.head()
```
